```python
import tensorflow as tf
from tensorflow.keras import layers, Input, Model

# 1. Definisi Input (4 fitur antropometri) [cite: 93-97]
inputs = Input(shape=(4,), name="antropometri_input")

# 2. Hidden Layers sesuai rancangan [cite: 19-23]
x = layers.Dense(64, activation='relu')(inputs)
x = layers.Dropout(0.2)(x)
x = layers.Dense(32, activation='relu')(x)
x = layers.Dropout(0.2)(x)
x = layers.Dense(16, activation='relu')(x)

# 3. Output Layer (Prediksi Stunting) [cite: 24, 152-153]
outputs = layers.Dense(1, activation='sigmoid')(x)

# 4. Inisialisasi & Compile Model [cite: 158-160]
model = Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Tampilkan ringkasan model untuk bukti progres di GitHub
model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Input, Model

# Input sesuai dokumen: umur, BB, TB, jenis kelamin [cite: 6-9]
inputs = Input(shape=(4,), name="antropometri_input")

# Arsitektur sesuai desain rekan kamu [cite: 19-24]
x = layers.Dense(64, activation='relu')(inputs)
x = layers.Dropout(0.2)(x)
x = layers.Dense(32, activation='relu')(x)
x = layers.Dropout(0.2)(x)
x = layers.Dense(16, activation='relu')(x)

# Output Layer untuk prediksi stunting [cite: 24, 29]
outputs = layers.Dense(1, activation='sigmoid')(x)

model = Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.summary()
```
